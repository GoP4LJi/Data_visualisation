```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../Data/virat_rohit_linechart.csv')
```

```python
df.dropna(how="all", inplace=True, axis=1)
```

```python
df
```

```python
plt.plot(df['Year'], df['Virat_Runs'], label='Virat Kohli', color="Blue")
plt.plot(df['Year'], df['Rohit_Runs'], label='Rohit Sharma', color="Red", linestyle='dashed')
plt.legend()
plt.xlabel('Year')
plt.ylabel('Runs')
```

```python
figure , axis = plt.subplots(2,2, figsize=(6,6), sharex=True, sharey=True)
axis[0,0].plot(df['Year'], df['Virat_Runs'], label='Virat Kohli', color="Blue" )
axis[0,0].set_title('Virat Kohli Runs')
axis[0,1].plot(df['Year'], df['Rohit_Runs'], label='Rohit Sharma', color="Red")
axis[0,1].set_title('Rohit Sharma Runs')
axis[1,0].scatter(df['Year'], df['Virat_Runs'], label='Virat Kohli', color="Blue")
axis[1,0].set_title('Virat Kohli Runs Scatter')
axis[1,1].scatter(df['Year'], df['Rohit_Runs'], label='Rohit Sharma', color="Red")
axis[1,1].set_title('Rohit Sharma Runs Scatter')
figure.suptitle('Virat Kohli and Rohit Sharma Runs Over the Years')
plt.tight_layout()
```

```python
plt.subplot(2,2,1)
plt.plot(df['Year'], df['Virat_Runs'], label='Virat Kohli', color="Blue" )
plt.title('Virat Kohli Runs')
plt.subplot(2,2,2)
plt.plot(df['Year'], df['Rohit_Runs'], label='Rohit Sharma', color="Red")
plt.title('Rohit Sharma Runs')
```

```python
tip2 = df.groupby('day')[['total_bill','tip']].sum().reset_index()
tip2.plot(kind='bar', x='day', y=['total_bill', 'tip'])
```

```python
X = np.arange(len(tip2['day']))
plt.bar(x=X, height=tip2['tip'], width=0.25, label='Tip')
plt.bar(x=X + 0.25, height=tip2['total_bill'], width=0.25, label='Total Bill')
```

```python
plt.bar(x=X, height=tip2['tip'], width=0.25, label='Tip', bottom=tip2['total_bill'])
plt.bar(x=X, height=tip2['total_bill'], width=0.25, label='Total Bill')
```
