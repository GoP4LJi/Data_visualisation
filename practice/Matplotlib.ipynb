```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../Data/virat_rohit_linechart.csv')
```

```python
df.dropna(how="all", inplace=True, axis=1)
```

```python
df
```

```python
plt.plot(df['Year'], df['Virat_Runs'], label='Virat Kohli', color="Blue")
plt.plot(df['Year'], df['Rohit_Runs'], label='Rohit Sharma', color="Red", linestyle='dashed')
plt.legend()
plt.xlabel('Year')
plt.ylabel('Runs')
```

```python
figure , axis = plt.subplots(2,2, figsize=(6,6), sharex=True, sharey=True)
axis[0,0].plot(df['Year'], df['Virat_Runs'], label='Virat Kohli', color="Blue" )
axis[0,0].set_title('Virat Kohli Runs')
axis[0,1].plot(df['Year'], df['Rohit_Runs'], label='Rohit Sharma', color="Red")
axis[0,1].set_title('Rohit Sharma Runs')
axis[1,0].scatter(df['Year'], df['Virat_Runs'], label='Virat Kohli', color="Blue")
axis[1,0].set_title('Virat Kohli Runs Scatter')
axis[1,1].scatter(df['Year'], df['Rohit_Runs'], label='Rohit Sharma', color="Red")
axis[1,1].set_title('Rohit Sharma Runs Scatter')
figure.suptitle('Virat Kohli and Rohit Sharma Runs Over the Years')
plt.tight_layout()
```

```python
plt.subplot(2,2,1)
plt.plot(df['Year'], df['Virat_Runs'], label='Virat Kohli', color="Blue" )
plt.title('Virat Kohli Runs')
plt.subplot(2,2,2)
plt.plot(df['Year'], df['Rohit_Runs'], label='Rohit Sharma', color="Red")
plt.title('Rohit Sharma Runs')
```

```python
df = sns.load_dataset('tips')
df.head()
```

```python
plt.scatter(df['total_bill'], df['tip'])
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('Scatter Plot of Total Bill vs Tip')
```

```python
sns.scatterplot(x = df['total_bill'], y = df['tip'] , hue = df['sex'])
```

```python
plt.scatter(df['total_bill'], df['tip'], c=df['sex'].map({'Male': 'blue', 'Female': 'red'}))
```

```python
males = df[df['sex'] == 'Male']
females = df[df['sex'] == 'Female']
plt.scatter(males['total_bill'], males['tip'], color='blue', label='Male')
plt.scatter(females['total_bill'], females['tip'], color='red', label='Female')
plt.legend()
```

```python
sns.scatterplot(data=df, x='total_bill', y='tip', hue='sex',size='size')
```

```python
sns.relplot(data=df, x='total_bill', y='tip', hue='sex',size='size', kind='scatter', col="smoker" , row="sex")
# fecesiting 
# rel can create two types of plots scatter and line plot
```

```python

sns.barplot(data=df, x='day', y='total_bill')
```
