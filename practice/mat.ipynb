```python
import matplotlib.pyplot as plt
import seaborn as sns
import  numpy as np 
```

```python
x = np.linspace(0, 10, 100)
y = x**2
```

```python

plt.plot(x, y)
plt.title("Time")
plt.xlabel("x-axis")
plt.ylabel('y-axis')
```

# Subplots

```python
plt.subplot(2,2,1) 
plt.plot(y,x)
plt.subplot(2,2,4) 
plt.plot(x**2,x**2) 
plt.subplot(2,2,2)
plt.plot(y**2,y)
plt.subplot(2,2,3) 
plt.plot(x,y)
```

# Object Oriented Approach

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,0.8,0.8]) # this is the main axes, note the first two numbers are the x and y coordinates of the lower left corner of the axes, and the last two numbers are the width and height of the axes
axis1.plot(x,y)
axis2 = fig.add_axes([0.2,0.5,0.4,0.3]) # this is the inset axes
axis2.plot(y,x)
```
